```python
import kagglehub

# Download latest version : https://www.kaggle.com/datasets/erogluegemen/airline-passengers
path = kagglehub.dataset_download("erogluegemen/airline-passengers")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# --- Charger le dataset ---
# Télécharge "AirPassengers.csv" depuis Kaggle ou UCI et place-le dans le dossier courant
df = pd.read_csv("data/airline-passengers.csv", parse_dates=["month"], index_col="month")

# --- Aperçu du dataset ---
print(df.head())
print("\nDataset shape:", df.shape)
```

```python

# --- Info et stats de base ---
print("\nData types:")
print(df.dtypes)

print("\nSummary statistics:")
print(df.describe())
```

```python
# --- Tracer la série temporelle ---
plt.figure(figsize=(12, 5))
plt.plot(df.index, df["total_passengers"], label="Passengers")
plt.title("Monthly Airline Passengers (1949-1960)")
plt.xlabel("Date")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()
```

```python
# --- Zoom sur une période ---
df["1955":"1957"].plot(figsize=(12,4), legend=False, title="Zoom on 1955-1957")
plt.ylabel("Passengers")
plt.show()
```

```python
# --- Exemple: moyenne mobile (rolling mean) ---
df["rolling_mean_12"] = df["total_passengers"].rolling(window=12).mean()

plt.figure(figsize=(12,5))
plt.plot(df.index, df["total_passengers"], label="Passengers")
plt.plot(df.index, df["rolling_mean_12"], label="12-month rolling mean", color="red")
plt.title("Passengers with 12-Month Rolling Mean")
plt.legend()
plt.show()
```
