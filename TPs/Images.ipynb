```python
import kagglehub

# Download latest version : https://www.kaggle.com/datasets/msarmi9/food101tiny
path = kagglehub.dataset_download("msarmi9/food101tiny")

print("Path to dataset files:", path)
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import random

data_path = os.path.join('data', 'image', 'food-101-tiny', 'train', 'apple_pie')
```

```python
# --- Définir le chemin vers le dossier contenant les images ---
data_dir = data_path

# --- Lister quelques fichiers ---
image_files = os.listdir(data_dir)
print("Number of images in folder:", len(image_files))
print("First 5 images:", image_files[:5])
```

```python
# --- Afficher une seule image ---
img_path = os.path.join(data_dir, image_files[0])
img = mpimg.imread(img_path)

plt.imshow(img)
plt.axis("off")   # masque les axes
plt.title("Example image")
plt.show()
```

```python
# --- Afficher plusieurs images au hasard ---
n = 5
sample_files = random.sample(image_files, n)

plt.figure(figsize=(15, 5))
for i, file in enumerate(sample_files):
    img = mpimg.imread(os.path.join(data_dir, file))
    plt.subplot(1, n, i+1)
    plt.imshow(img)
    plt.axis("off")
    plt.title(file[:10])  # affiche le nom tronqué
plt.show()
```

```python
# --- Vérifier les dimensions d’une image --- 
print(img.shape)  # (height, width, channels)
```

```python
# --- Convertir en array NumPy --- 
import numpy as np
print("Min pixel:", np.min(img))
print("Max pixel:", np.max(img))
print("Mean pixel value:", np.mean(img))
```

```python
# Sans rouge
img_no_red = img.copy()
img_no_red[:, :, 0] = 0

# Sans vert
img_no_green = img.copy()
img_no_green[:, :, 1] = 0

# Sans bleu
img_no_blue = img.copy()
img_no_blue[:, :, 2] = 0
```

```python
plt.imshow(img_no_blue)
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img_box = img.copy()

# coordonnées du rectangle
x1, y1 = 20, 40
x2, y2 = 180, 160

cv2.rectangle(
    img_box,
    (x1, y1),
    (x2, y2),
    (1, 0, 0),   # rouge si image normalisée 0-1
    4
)

plt.imshow(img_box)
plt.axis("off")
plt.show()
```
